```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("hotel_bookings.csv")

# 1. Missing values count and percentage
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[missing_summary["Missing Count"] > 0]
missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

print("Missing Values Summary:")
display(missing_summary)

# 2. Identify columns with very high missingness
print("Columns with more than 50% missing values:")
high_missing = missing_summary[missing_summary["Missing Percentage"] > 50]
display(high_missing)

# 3. Remove data leakage columns
leakage_columns = ["reservation_status", "reservation_status_date"]

print("Leakage columns:")
print(leakage_columns)

df = df.drop(columns=leakage_columns)

print("\nShape after removing leakage columns:")
print(df.shape)

# 4. Check selected numerical features for outliers using IQR
selected_numerical = [
    "lead_time",
    "adr",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies"
]

for column in selected_numerical:
    if column in df.columns:
        plt.figure(figsize=(8, 4))
        plt.boxplot(df[column].dropna())
        plt.title(f"Boxplot of {column}")
        plt.ylabel(column)
        plt.show()

# 5. Count outliers using IQR
outlier_summary = []

for column in selected_numerical:
    if column in df.columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ]

        outlier_summary.append({
            "Column": column,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Count": len(outliers)
        })

outlier_summary = pd.DataFrame(outlier_summary)

print("Outlier Summary:")
display(outlier_summary)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder

y = df["is_canceled"]
X = df.drop(columns=["is_canceled"])

numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# Use a representative training sample for KNN fitting
knn_sample_size = min(10000, len(X_train))

X_knn_sample = X_train[numerical_columns].sample(
    n=knn_sample_size,
    random_state=42
)

# Pipeline A numerical preprocessing
numerical_pipeline_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

# Pipeline B numerical preprocessing
numerical_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=True
    ))
])

# Fit KNN imputers using training sample
print("Fitting numerical Pipeline A...")
numerical_pipeline_A.fit(X_knn_sample)

print("Fitting numerical Pipeline B...")
numerical_pipeline_B.fit(X_knn_sample)

# Transform numerical data
print("Transforming numerical data...")

X_train_num_A = numerical_pipeline_A.transform(
    X_train[numerical_columns]
)

X_test_num_A = numerical_pipeline_A.transform(
    X_test[numerical_columns]
)

X_train_num_B = numerical_pipeline_B.transform(
    X_train[numerical_columns]
)

X_test_num_B = numerical_pipeline_B.transform(
    X_test[numerical_columns]
)

# Fit categorical preprocessing only on training data
print("Processing categorical data...")

X_train_cat = categorical_pipeline.fit_transform(
    X_train[categorical_columns]
)

X_test_cat = categorical_pipeline.transform(
    X_test[categorical_columns]
)

# Combine numerical and categorical features
from scipy.sparse import hstack

X_train_A = hstack([
    X_train_num_A,
    X_train_cat
]).tocsr()

X_test_A = hstack([
    X_test_num_A,
    X_test_cat
]).tocsr()

X_train_B = hstack([
    X_train_num_B,
    X_train_cat
]).tocsr()

X_test_B = hstack([
    X_test_num_B,
    X_test_cat
]).tocsr()

print("\nTask 3 completed successfully.")

print("Pipeline A training shape:", X_train_A.shape)
print("Pipeline A testing shape:", X_test_A.shape)

print("Pipeline B training shape:", X_train_B.shape)
print("Pipeline B testing shape:", X_test_B.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

# Logistic Regression - Pipeline A (StandardScaler)
print("Training Logistic Regression - StandardScaler...")
logistic_A = LogisticRegression(max_iter=1000)
logistic_A.fit(X_train_A, y_train)

# Logistic Regression - Pipeline B (MinMaxScaler)
print("Training Logistic Regression - MinMaxScaler...")
logistic_B = LogisticRegression(max_iter=1000)
logistic_B.fit(X_train_B, y_train)

# Decision Tree - Pipeline A (StandardScaler)
print("Training Decision Tree - StandardScaler...")
tree_A = DecisionTreeClassifier(random_state=42)
tree_A.fit(X_train_A, y_train)

# Decision Tree - Pipeline B (MinMaxScaler)
print("Training Decision Tree - MinMaxScaler...")
tree_B = DecisionTreeClassifier(random_state=42)
tree_B.fit(X_train_B, y_train)

print("\nAll four model-pipeline combinations trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import pandas as pd
import matplotlib.pyplot as plt

# Store the four models
models = {
    "Logistic Regression - StandardScaler": (
        logistic_A, X_train_A, X_test_A
    ),
    "Logistic Regression - MinMaxScaler": (
        logistic_B, X_train_B, X_test_B
    ),
    "Decision Tree - StandardScaler": (
        tree_A, X_train_A, X_test_A
    ),
    "Decision Tree - MinMaxScaler": (
        tree_B, X_train_B, X_test_B
    )
}

results = []

# Evaluate all four models
for name, (model, Xtr, Xte) in models.items():

    y_train_pred = model.predict(Xtr)
    y_test_pred = model.predict(Xte)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(
        y_test, y_test_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_test_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_test_pred, zero_division=0
    )

    results.append({
        "Model": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

# Create comparison table
results_df = pd.DataFrame(results)

print("FINAL MODEL COMPARISON")
display(results_df.round(4))

# --------------------------------------------------
# Best Logistic Regression
# --------------------------------------------------

logistic_results = results_df[
    results_df["Model"].str.startswith("Logistic Regression")
]

best_logistic_name = logistic_results.loc[
    logistic_results["F1-Score"].idxmax(),
    "Model"
]

best_logistic_model = models[best_logistic_name][0]
best_logistic_X_test = models[best_logistic_name][2]

print("\nBest Logistic Regression:")
print(best_logistic_name)

# Predictions
y_pred_logistic = best_logistic_model.predict(
    best_logistic_X_test
)

# Confusion matrix
cm_logistic = confusion_matrix(
    y_test,
    y_pred_logistic
)

print("\nLogistic Regression Confusion Matrix:")
print(cm_logistic)

ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Canceled", "Canceled"]
).plot()

plt.title(
    f"Confusion Matrix - {best_logistic_name}"
)
plt.show()

# --------------------------------------------------
# Best Decision Tree
# --------------------------------------------------

tree_results = results_df[
    results_df["Model"].str.startswith("Decision Tree")
]

best_tree_name = tree_results.loc[
    tree_results["F1-Score"].idxmax(),
    "Model"
]

best_tree_model = models[best_tree_name][0]
best_tree_X_test = models[best_tree_name][2]

print("\nBest Decision Tree:")
print(best_tree_name)

# Predictions
y_pred_tree = best_tree_model.predict(
    best_tree_X_test
)

# Confusion matrix
cm_tree = confusion_matrix(
    y_test,
    y_pred_tree
)

print("\nDecision Tree Confusion Matrix:")
print(cm_tree)

ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Not Canceled", "Canceled"]
).plot()

plt.title(
    f"Confusion Matrix - {best_tree_name}"
)
plt.show()

# --------------------------------------------------
# Overfitting analysis
# --------------------------------------------------

results_df["Train-Test Difference"] = (
    results_df["Training Accuracy"]
    - results_df["Testing Accuracy"]
)

print("\nTRAIN-TEST PERFORMANCE DIFFERENCE")

display(
    results_df[
        [
            "Model",
            "Training Accuracy",
            "Testing Accuracy",
            "Train-Test Difference"
        ]
    ].round(4)
)
```

```python
# Best overall model based on F1-score
best_overall = results_df.loc[
    results_df["F1-Score"].idxmax()
]

print("BEST OVERALL MODEL")
print(best_overall["Model"])
print(f"F1-Score: {best_overall['F1-Score']:.4f}")
print(f"Testing Accuracy: {best_overall['Testing Accuracy']:.4f}")

# Logistic Regression comparison
logistic_comparison = results_df[
    results_df["Model"].str.startswith("Logistic Regression")
]

print("\nLOGISTIC REGRESSION COMPARISON")
display(logistic_comparison.round(4))

# Decision Tree comparison
tree_comparison = results_df[
    results_df["Model"].str.startswith("Decision Tree")
]

print("\nDECISION TREE COMPARISON")
display(tree_comparison.round(4))

# Best Logistic Regression
best_logistic = logistic_comparison.loc[
    logistic_comparison["F1-Score"].idxmax()
]

print("\nBEST LOGISTIC REGRESSION")
print(best_logistic["Model"])
print(f"F1-Score: {best_logistic['F1-Score']:.4f}")

# Best Decision Tree
best_tree = tree_comparison.loc[
    tree_comparison["F1-Score"].idxmax()
]

print("\nBEST DECISION TREE")
print(best_tree["Model"])
print(f"F1-Score: {best_tree['F1-Score']:.4f}")

# Overfitting comparison
print("\nTRAIN-TEST PERFORMANCE DIFFERENCE")
display(
    results_df[
        [
            "Model",
            "Training Accuracy",
            "Testing Accuracy",
            "Train-Test Difference"
        ]
    ].round(4)
)
```
