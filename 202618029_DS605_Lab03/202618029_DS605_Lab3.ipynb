```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("hotel_bookings.csv")

print("Import all Libraries ans CSV")
```

```python
print("First 5 rows:")
display(df.head())

print("Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Description:")
display(df.describe())

print("\nData Types:")
print(df.dtypes)

print("\nClass Distribution of is_canceled:")
print(df["is_canceled"].value_counts())

print("\nClass Distribution in Percentage:")
print(df["is_canceled"].value_counts(normalize=True) * 100)

y = df["is_canceled"]
X = df.drop(columns=["is_canceled"])

numerical_columns = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_columns = X.select_dtypes(include=["object"]).columns.tolist()

print("\nTarget variable (y):")
print(y.name)

print("\nNumerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
# 1. Missing values count and percentage
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[missing_summary["Missing Count"] > 0]
missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

print("Missing Values Summary:")
display(missing_summary)

# 2. Identify columns with very high missingness
print("Columns with more than 50% missing values:")
high_missing = missing_summary[missing_summary["Missing Percentage"] > 50]
display(high_missing)

# 3. Remove data leakage columns
leakage_columns = ["reservation_status", "reservation_status_date"]

print("Leakage columns:")
print(leakage_columns)

df = df.drop(columns=leakage_columns)

print("\nShape after removing leakage columns:")
print(df.shape)

# 4. Check selected numerical features for outliers using IQR
selected_numerical = [
    "lead_time",
    "adr",
    "stays_in_weekend_nights",
    "stays_in_week_nights",
    "adults",
    "children",
    "babies"
]

for column in selected_numerical:
    if column in df.columns:
        plt.figure(figsize=(8, 4))
        plt.boxplot(df[column].dropna())
        plt.title(f"Boxplot of {column}")
        plt.ylabel(column)
        plt.show()

# 5. Count outliers using IQR
outlier_summary = []

for column in selected_numerical:
    if column in df.columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = df[
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ]

        outlier_summary.append({
            "Column": column,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Count": len(outliers)
        })

outlier_summary = pd.DataFrame(outlier_summary)

print("Outlier Summary:")
display(outlier_summary)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import KNNImputer, SimpleImputer
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
```

```python
# Load dataset
df = pd.read_csv("hotel_bookings.csv")

# Remove data leakage columns
df = df.drop(columns=["reservation_status", "reservation_status_date"])

# Target variable
y = df["is_canceled"]

# Features
X = df.drop(columns=["is_canceled"])

# Identify numerical and categorical columns
numerical_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_columns = X.select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("\nTraining data shape:")
print(X_train.shape)

print("\nTesting data shape:")
print(X_test.shape)

# Numerical preprocessing for Pipeline A
numerical_pipeline_A = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", StandardScaler())
])

# Numerical preprocessing for Pipeline B
numerical_pipeline_B = Pipeline([
    ("imputer", KNNImputer(n_neighbors=5)),
    ("scaler", MinMaxScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Pipeline A
preprocessor_A = ColumnTransformer([
    ("numerical", numerical_pipeline_A, numerical_columns),
    ("categorical", categorical_pipeline, categorical_columns)
])

# Pipeline B
preprocessor_B = ColumnTransformer([
    ("numerical", numerical_pipeline_B, numerical_columns),
    ("categorical", categorical_pipeline, categorical_columns)
])

print("\nPipeline A:")
print(preprocessor_A)

print("\nPipeline B:")
print(preprocessor_B)
```

```python
# Logistic Regression with Pipeline A
logistic_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", LogisticRegression(max_iter=1000))
])

# Logistic Regression with Pipeline B
logistic_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", LogisticRegression(max_iter=1000))
])

# Decision Tree with Pipeline A
tree_A = Pipeline([
    ("preprocessor", preprocessor_A),
    ("model", DecisionTreeClassifier(random_state=42))
])

# Decision Tree with Pipeline B
tree_B = Pipeline([
    ("preprocessor", preprocessor_B),
    ("model", DecisionTreeClassifier(random_state=42))
])

# Train all four models
logistic_A.fit(X_train, y_train)
logistic_B.fit(X_train, y_train)

tree_A.fit(X_train, y_train)
tree_B.fit(X_train, y_train)

print("All four model-pipeline combinations trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import pandas as pd
import matplotlib.pyplot as plt

# Store all trained models
models = {
    "Logistic Regression - StandardScaler": logistic_A,
    "Logistic Regression - MinMaxScaler": logistic_B,
    "Decision Tree - StandardScaler": tree_A,
    "Decision Tree - MinMaxScaler": tree_B
}

results = []

# Evaluate all four models
for name, model in models.items():

    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)

    train_accuracy = accuracy_score(y_train, y_train_pred)
    test_accuracy = accuracy_score(y_test, y_test_pred)
    precision = precision_score(y_test, y_test_pred, zero_division=0)
    recall = recall_score(y_test, y_test_pred, zero_division=0)
    f1 = f1_score(y_test, y_test_pred, zero_division=0)

    results.append({
        "Model": name,
        "Training Accuracy": train_accuracy,
        "Testing Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1
    })

# Create comparison table
results_df = pd.DataFrame(results)

print("Final Model Comparison:")
display(results_df.round(4))

# Find best Logistic Regression model based on F1-score
logistic_results = results_df[
    results_df["Model"].str.startswith("Logistic Regression")
]

best_logistic_name = logistic_results.loc[
    logistic_results["F1-Score"].idxmax(), "Model"
]

best_logistic_model = models[best_logistic_name]

# Find best Decision Tree model based on F1-score
tree_results = results_df[
    results_df["Model"].str.startswith("Decision Tree")
]

best_tree_name = tree_results.loc[
    tree_results["F1-Score"].idxmax(), "Model"
]

best_tree_model = models[best_tree_name]

print("\nBest Logistic Regression:")
print(best_logistic_name)

print("\nBest Decision Tree:")
print(best_tree_name)

# Confusion Matrix - Best Logistic Regression
y_pred_logistic = best_logistic_model.predict(X_test)

cm_logistic = confusion_matrix(y_test, y_pred_logistic)

print("\nLogistic Regression Confusion Matrix:")
print(cm_logistic)

ConfusionMatrixDisplay(
    confusion_matrix=cm_logistic,
    display_labels=["Not Canceled", "Canceled"]
).plot()

plt.title(f"Confusion Matrix - {best_logistic_name}")
plt.show()

# Confusion Matrix - Best Decision Tree
y_pred_tree = best_tree_model.predict(X_test)

cm_tree = confusion_matrix(y_test, y_pred_tree)

print("\nDecision Tree Confusion Matrix:")
print(cm_tree)

ConfusionMatrixDisplay(
    confusion_matrix=cm_tree,
    display_labels=["Not Canceled", "Canceled"]
).plot()

plt.title(f"Confusion Matrix - {best_tree_name}")
plt.show()

# Overfitting analysis
results_df["Train-Test Difference"] = (
    results_df["Training Accuracy"] -
    results_df["Testing Accuracy"]
)

print("\nTrain-Test Performance Difference:")
display(
    results_df[
        ["Model", "Training Accuracy", "Testing Accuracy", "Train-Test Difference"]
    ].round(4)
)
```

```python
print("Best Overall Model:")
best_model = results_df.loc[results_df["F1-Score"].idxmax()]
display(best_model.to_frame().T.round(4))

print("\nComparison of Logistic Regression:")
display(
    results_df[
        results_df["Model"].str.startswith("Logistic Regression")
    ].round(4)
)

print("\nComparison of Decision Tree:")
display(
    results_df[
        results_df["Model"].str.startswith("Decision Tree")
    ].round(4)
)
```
