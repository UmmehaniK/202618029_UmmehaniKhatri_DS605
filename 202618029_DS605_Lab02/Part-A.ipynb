```python
A = np.array([
    [1, 2],
    [3, 4]
])

print("Matrix A:")
print(A)
B = np.array([
    [5, 6],
    [7, 8]
])
print(B)
print("Matrix B: ",A+B)
print("Element-wise multiplication: ",A*B)
print("Matrix multiplication: ",A @ B)
print("Transpose of A: ",A.T)

np.linalg.det(A)
det_A = np.linalg.det(A)
print("Determinant of A:", det_A)

inverse_A = np.linalg.inv(A)
print("Inverse of A:")
print(inverse_A)

result = A @ inverse_A
print("A @ A^-1:")
print(result)

print("Is it identity matrix?", np.allclose(result, np.eye(2)))
np.allclose(result, np.eye(2))
```

```python
mean = 50
std = 10

print("Chosen Mean:", mean)
print("Chosen Standard Deviation:", std)

np.random.seed(42)

data = np.random.normal(loc=mean, scale=std, size=1000)

print(data)
print("Number of values:", len(data))
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

print("Sample Mean:", sample_mean)
sample_mean = np.mean(data)

print("Sample Mean:", sample_mean)
np.std(data, ddof=1)
print("Chosen Mean:", mean)
print("Sample Mean:", sample_mean)

print()

print("Chosen Standard Deviation:", std)
print("Sample Standard Deviation:", sample_std)
plt.hist(data, bins=30)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")
plt.show()
```
